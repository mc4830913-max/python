```python

import pandas as pd
import plotly.express as px

# Ensure datetime
df["last_updated"] = pd.to_datetime(df["last_updated"])

# Extract year
df["last_updated_year"] = df["last_updated"].dt.year

# Count tags by year
plot_df = (
    df.groupby(["last_updated_year", "clean_tags"])
      .size()
      .reset_index(name="count")
)

# Keep top 10 tags for each year
plot_df = (
    plot_df.sort_values(["last_updated_year", "count"], ascending=[True, False])
           .groupby("last_updated_year")
           .head(25)
)

# Interactive animated bar chart
fig = px.bar(
    plot_df,
    x="clean_tags",
    y="count",
    animation_frame="last_updated_year",  # Slider shows years only
    color_discrete_sequence=["#9B59B6"]
)

# Purple borders
fig.update_traces(
    marker=dict(
        line=dict(color="#C77DFF", width=2)
    )
)

# Black theme
fig.update_layout(
    title="Top 10 Clean Tags by Year",
    plot_bgcolor="black",
    paper_bgcolor="black",
    font=dict(color="white"),
    xaxis=dict(
        title="Clean Tags",
        tickangle=-45,
        gridcolor="#6A0DAD",
        linecolor="#C77DFF",
        color="white"
    ),
    yaxis=dict(
        title="Count",
        gridcolor="#6A0DAD",
        linecolor="#C77DFF",
        color="white"
    ),
    height=650
)

fig.show()
```

```python
import plotly.express as px

# Count tags
tag_counts = (
    df["clean_tags"]
    .value_counts()
    .reset_index().head(10)
)

tag_counts.columns = ["clean_tags", "count"]

# Purple color palette
purple_colors = [
    "#4B0082", "#5D3FD3", "#6A0DAD", "#7B2CBF", "#8A2BE2",
    "#9B59B6", "#A855F7", "#B266FF", "#C77DFF", "#D8B4FE",
    "#E0AAFF", "#F3E8FF"
]

# Interactive Pie Chart
fig = px.pie(
    tag_counts,
    names="clean_tags",
    values="count",
    color_discrete_sequence=purple_colors
)

# Purple borders and hover
fig.update_traces(
    textinfo="percent+label",
    hovertemplate="<b>%{label}</b><br>Count: %{value}<br>Percentage: %{percent}<extra></extra>",
    marker=dict(
        line=dict(color="#C77DFF", width=2)
    )
)

# Black theme
fig.update_layout(
    title="Distribution of Clean Tags",
    plot_bgcolor="black",
    paper_bgcolor="black",
    font=dict(color="white", size=14),
    legend=dict(
        bgcolor="black",
        font=dict(color="white")
    ),
    height=700,
    width=900
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px

# Convert dictionary to DataFrame
plot_df = pd.DataFrame(
    dct.items(),
    columns=["clean_tags", "frequency"]
).sort_values("frequency", ascending=False)

# Purple color palette
purple_colors = [
    "#4B0082", "#5D3FD3", "#6A0DAD", "#7B2CBF", "#8A2BE2",
    "#9B59B6", "#A855F7", "#B266FF", "#C77DFF", "#D8B4FE",
    "#E0AAFF", "#F3E8FF"
]

# Interactive Donut Chart
fig = px.pie(
    plot_df,
    names="clean_tags",
    values="frequency",
    hole=0.55,  # Creates the donut
    color_discrete_sequence=purple_colors
)

# Purple borders and hover
fig.update_traces(
    textinfo="label+percent",
    marker=dict(
        line=dict(color="#C77DFF", width=2)
    ),
    hovertemplate="<b>%{label}</b><br>Appeared in Top 10: %{value} years<br>%{percent}<extra></extra>"
)

# Black theme
fig.update_layout(
    title="Frequency of Tags Appearing in the Yearly Top 10 (2010–2025)",
    plot_bgcolor="black",
    paper_bgcolor="black",
    font=dict(color="white", size=14),
    legend=dict(
        bgcolor="black",
        font=dict(color="white"),
        title="Tags"
    ),
    height=750,
    width=950
)

fig.show()
```

```python

import plotly.express as px

# Count occurrences
plot_df = (
    df.groupby(["broader_category", "clean_tags"])
      .size()
      .reset_index(name="count").head(10)
)

# Purple color palette
purple_colors = [
    "#4B0082", "#5D3FD3", "#6A0DAD", "#7B2CBF",
    "#8A2BE2", "#9B59B6", "#A855F7",
    "#B266FF", "#C77DFF", "#D8B4FE",
    "#E0AAFF", "#F3E8FF"
]

# Sunburst Chart
fig = px.sunburst(
    plot_df,
    path=["broader_category", "clean_tags"],
    values="count",
    color="broader_category",
    color_discrete_sequence=purple_colors
)

# Purple borders
fig.update_traces(
    marker=dict(
        line=dict(color="#C77DFF", width=2)
    ),
    hovertemplate="<b>%{label}</b><br>Count: %{value}<extra></extra>"
)

# Black theme
fig.update_layout(
    title="Broader Categories and Their Tags",
    paper_bgcolor="black",
    plot_bgcolor="black",
    font=dict(color="white", size=14),
    height=800,
    width=1000,
    margin=dict(t=60, l=20, r=20, b=20)
)

fig.show()
```
